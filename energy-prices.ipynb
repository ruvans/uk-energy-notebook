```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import ArrowStyle

df = pd.read_csv(r"C:\Users\ruthe\Documents Local\data projects\uk_energy_prices_annual.csv")
df.head()
```

```python
fig, ax = plt.subplots(figsize=(10,6))

# Plot the real prices
ax.plot(df['Year'], df['Real price indices: Electricity [Note 3]'], label='Electricity', color='gold')
ax.plot(df['Year'], df['Real price indices: Gas [Note 3]'], label='Gas', color='steelblue')
ax.plot(df['Year'], df['Real price indices: Liquid fuels [Note 3]'], label='Oil', color='firebrick')
# graph looks too busy, we'll hide the lesser important energies
# motor fuel is rises and falls similarly to oil, solid fuels isn't affected much by any events.
#ax.plot(df['Year'], df['Real price indices: Motor fuel & oil\n[Note 2 ,3]'], label='Motor oil', color='darkorange')
#ax.plot(df['Year'], df['Real price indices: Solid fuels [Note 3]'], label='Solid fuels', color='brown')

# Add title with custom font size and weight
ax.set_title("Energy prices in the UK 1990-2025", size=16, weight="bold")

# Customize x and y labels
ax.set_xlabel("Year", size=12)
ax.set_ylabel("Price", size=12)

ax.legend()

arrows=dict(arrowstyle='->')

ax.annotate('US-led invasion of Iraq', xy=(0.418, 0.3), xycoords='figure fraction', xytext=(-220, 60), textcoords='offset pixels',horizontalalignment='left', verticalalignment='top',
            fontsize=12,  arrowprops=arrows)
ax.annotate('covid-19 lockdowns', xy=(0.82, 0.306), xycoords='figure fraction', xytext=(-100,-60), textcoords='offset pixels',horizontalalignment='left', verticalalignment='top',
            fontsize=12,  arrowprops=arrows)
ax.annotate('Russia invades Ukraine', xy=(0.864, 0.88), xycoords='figure fraction', xytext=(-260, 0), textcoords='offset pixels',horizontalalignment='left', verticalalignment='top',
            fontsize=12, arrowprops=arrows)
# both the 2008 peak and the 2009 trough are due to the global financial crisis of 2008
# https://commonslibrary.parliament.uk/research-briefings/sn02106/
arrowstyle=ArrowStyle.Fancy(head_length=.4, head_width=.4, tail_width=.4)
ax.annotate('2008 financial crisis', xy=(0.52, 0.62), xycoords='figure fraction', xytext=(-220, 0), textcoords='offset pixels',horizontalalignment='left', verticalalignment='top',
            fontsize=12,  arrowprops=arrows)

plt.tight_layout()
```
